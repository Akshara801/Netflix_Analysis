```python
# --------------------------------------------
# Netflix Dataset Analysis
# --------------------------------------------

# 📦 Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ✅ Display settings
pd.set_option('display.max_columns', None)
plt.style.use('seaborn-v0_8')

# --------------------------------------------
# 📂 Load the dataset
# --------------------------------------------
# Make sure 'netflix_titles.csv' is in the same folder as this notebook
df = pd.read_csv("netflix_titles.csv")

# --------------------------------------------
# 🔍 Basic Information
# --------------------------------------------
print("Dataset Shape:", df.shape)
print("\nDataset Info:")
print(df.info())

print("\nFirst 5 rows of data:")
print(df.head())

# --------------------------------------------
# 🧹 Data Cleaning
# --------------------------------------------
print("\nChecking for missing values:")
print(df.isnull().sum())

# Fill missing values where possible
df['country'].fillna('Unknown', inplace=True)
df['rating'].fillna('Not Rated', inplace=True)
df['duration'].fillna('Unknown', inplace=True)

# Drop duplicates if any
df.drop_duplicates(inplace=True)

# --------------------------------------------
# 📊 Exploratory Data Analysis (EDA)
# --------------------------------------------

# 🔸 Distribution of content type (Movies vs TV Shows)
plt.figure(figsize=(6,4))
sns.countplot(x='type', data=df, palette='coolwarm')
plt.title("Distribution of Content Type")
plt.xlabel("Type of Show")
plt.ylabel("Count")
plt.show()

# 🔸 Top 10 countries producing the most content
top_countries = df['country'].value_counts().head(10)
plt.figure(figsize=(8,4))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='magma')
plt.title("Top 10 Countries with Most Content on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()

# 🔸 Trend of content release over the years
df['release_year'] = df['release_year'].astype(str)
year_wise = df['release_year'].value_counts().sort_index()
plt.figure(figsize=(10,4))
sns.lineplot(x=year_wise.index, y=year_wise.values, marker='o', color='purple')
plt.title("Content Released Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()

# 🔸 Top 10 Genres (from 'listed_in' column)
genres = df['listed_in'].dropna().apply(lambda x: x.split(',')).explode()
top_genres = genres.value_counts().head(10)
plt.figure(figsize=(8,4))
sns.barplot(x=top_genres.values, y=top_genres.index, palette='plasma')
plt.title("Top 10 Popular Genres on Netflix")
plt.xlabel("Count")
plt.ylabel("Genre")
plt.show()

# --------------------------------------------
# 🎬 Deep Dive: Movies vs TV Shows
# --------------------------------------------
movie_df = df[df['type'] == 'Movie']
tv_df = df[df['type'] == 'TV Show']

print("\nAverage Movie Duration (min):")
print(movie_df['duration'].head())

print("\nTV Show Seasons:")
print(tv_df['duration'].value_counts().head())

# --------------------------------------------
# 🔎 Word Cloud of Directors (optional if wordcloud installed)
# --------------------------------------------
try:
    from wordcloud import WordCloud
    text = " ".join(str(i) for i in df['director'].dropna())
    plt.figure(figsize=(10,6))
    wc = WordCloud(background_color='black', max_words=100, colormap='Reds').generate(text)
    plt.imshow(wc, interpolation='bilinear')
    plt.axis('off')
    plt.title("Word Cloud of Directors")
    plt.show()
except:
    print("⚠️ Install wordcloud using: pip install wordcloud")

# --------------------------------------------
# ✅ Summary
# --------------------------------------------
print("\nSummary:")
print(f"Total Titles: {len(df)}")
print(f"Movies: {len(movie_df)} | TV Shows: {len(tv_df)}")
print(f"Data available from {df['release_year'].min()} to {df['release_year'].max()}")
```
